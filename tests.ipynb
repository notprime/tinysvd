```python
import torch
from svd import svd as solver
```

```python
A = torch.rand(5, 10)
```

```python
svd = solver.SVD(tolerance=1e-4, max_iterations=100000, seed=24)

svd.fit(A)
A_hat = svd.reconstruct()

print(torch.norm(A-A_hat))
```

```python
A, A_hat
```

```python
print(svd.U.shape)
print(svd.Sigma.shape)
print(svd.V.shape)
```

```python
I = svd.U.T @ svd.U

print(torch.linalg.norm(I - torch.eye(I.shape[0])))
```

```python
I = svd.V.T @ svd.V

print(torch.linalg.norm(I - torch.eye(I.shape[0])))
```

```python
for i in range(len(svd.singular_values)):
    lhs = A @ svd.V[:, i]
    rhs = svd.singular_values[i] * svd.U[:, i]

    print(torch.linalg.norm(lhs - rhs))
```

```python
U, S, Vh = torch.linalg.svd(A, full_matrices=False)
```

```python
print(torch.linalg.norm(S - svd.singular_values))
```
